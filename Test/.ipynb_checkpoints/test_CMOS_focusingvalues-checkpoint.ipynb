```python
def detection(data, nrows, ncols):#nrows, ncols=CMOSの領域分け
    bkg = Background2D(data, box_size=50, filter_size=3, bkg_estimator=MedianBackground())
    data_sub = data - bkg.background
    
    # -------------------------------
    # PSF（簡易ガウシアン）でスムージング
    psf_model = IntegratedGaussianPRF(sigma=3)  # 1.5 pixel
    y, x = np.mgrid[-5:6, -5:6]
    psf = psf_model(x, y)
    psf /= np.sum(psf)
    
    smoothed = convolve(data_sub, psf, mode='same')
    
    # -------------------------------
    # ピーク検出
    threshold = 5 * bkg.background_rms_median
    segm = detect_sources(smoothed, threshold, npixels=5)
    
    # -------------------------------
    # 検出した領域の重心座標
    cat = SourceCatalog(data_sub, segm)
    # 天体の数
    print(f"検出した天体の数: {len(cat)}")
    a_sigma = cat.semimajor_sigma.value  # pix
    b_sigma = cat.semiminor_sigma.value  # pix
    
    # 楕円ガウスの平均的な σ を採用
    sigma_mean = np.sqrt(a_sigma * b_sigma)
    
    # σ → FWHM (pix単位)
    fwhm_pix = sigma_mean * (2*np.sqrt(2*np.log(2)))
    
    # pixel scale (HSC: 0.168 arcsec/pix)
    pixel_scale = 0.168#0.23
    fwhm_arcsec = fwhm_pix * pixel_scale
    
    # 楕円のポジションアングル [rad] → deg に変換
    pa_rad = cat.orientation.to("rad").value  # rad
    pa_deg = np.degrees(pa_rad)
    det_smooth = pd.DataFrame({
        "ID": np.arange(1, len(cat) + 1),
        "xcentroid": cat.xcentroid,
        "ycentroid": cat.ycentroid,
        "segment_flux": cat.segment_flux,
        "a_sigma": a_sigma,
        "b_sigma": b_sigma,
        "fwhm_pix": fwhm_pix,
        "fwhm_arcsec": fwhm_arcsec,
        "b/a": b_sigma/a_sigma,
        "pa_deg": pa_deg,
    })
    ny, nx = data.shape
    
    # グリッドの分割数
    #nrows, ncols = 4, 5
    dy, dx = ny / nrows, nx / ncols
    
    # 各検出座標が属するグリッドを計算
    det_smooth["row_bin"] = (det_smooth["ycentroid"] // dy).astype(int)
    det_smooth["col_bin"] = (det_smooth["xcentroid"] // dx).astype(int)
    
    # 安全のため範囲外をクリップ
    det_smooth["row_bin"] = det_smooth["row_bin"].clip(0, nrows-1)
    det_smooth["col_bin"] = det_smooth["col_bin"].clip(0, ncols-1)
    
    # グリッド番号 (0〜9)
    det_smooth["grid_id"] = det_smooth["row_bin"] * ncols + det_smooth["col_bin"]
    return segm, cat, det_smooth
```

```python
nrows,ncols = 3,3
segments, catalogs, det_smooths = [], [], []
for image in image_list:
    segm, cat, det_smooth = detection(image, nrows,ncols)
    segments.append(segm)
    catalogs.append(cat) 
    det_smooths.append(det_smooth)
```

import numpy as np
from scipy.ndimage import gaussian_filter
import matplotlib.pyplot as plt

shape = image_list[0].shape
n_points = 1000
seeing_range = (1, 2)  # arcsec単位でのseeing範囲
mag_range = (16, 20)
F0 = 1e10
pixel_scale = 0.23  # HSC の場合

image = np.zeros(shape)

def seeing_to_sigma(seeing_arcsec, pixel_scale=pixel_scale):
    fwhm_pix = seeing_arcsec / pixel_scale
    sigma_pix = fwhm_pix / (2 * np.sqrt(2 * np.log(2)))
    return sigma_pix

# --------------------------
# 点を埋め込む
# --------------------------
artificial_star_list=[]
for _ in range(n_points):
    # ランダム座標
    x = np.random.randint(0, shape[1])
    y = np.random.randint(0, shape[0])
    
    # ランダム seeing → sigma
    seeing = np.random.uniform(*seeing_range)
    sigma = seeing_to_sigma(seeing)
    
    # ランダム等級 → フラックス
    mag = np.random.uniform(*mag_range)
    flux = F0 * 10**(-0.4 * mag)
    
    # パッチのサイズ
    radius = int(3*sigma)
    x_min, x_max = max(x-radius,0), min(x+radius+1, shape[1])
    y_min, y_max = max(y-radius,0), min(y+radius+1, shape[0])
    
    # 小パッチに delta 関数を置いてガウスぼかし
    tmp = np.zeros((y_max-y_min, x_max-x_min))
    tmp[y-y_min, x-x_min] = flux
    tmp = gaussian_filter(tmp, sigma=sigma)
    
    # 元画像に加算
    image[y_min:y_max, x_min:x_max] += tmp


# --------------------------
# 可視化
# --------------------------
plt.imshow(image, origin='lower', cmap='viridis', vmin=0, vmax=1)
plt.colorbar(label='Flux')
plt.show()

```python
def varconv(data,num_list):
    ver=[]
    tot=0
    for num in num_list:
        tot+=10**num
    for data_hol in data:
        #hol=[bin(int(data_hol[i]))[2:].zfill(17) for i in range(len(data_hol))]#2次元のデータを１次元にスライス&２進数に変換
        hol = [bin(int(data_hol[i]))[2:].zfill(17) if not np.isnan(data_hol[i]) else str(tot).zfill(17) for i in range(len(data_hol))]
        ver.append(hol)
    return ver
def read(data,mask_bit):
    bits=[]
    for hol in data:
        hol_new=[]
        for one_hol in hol:#one_holeは画素一つのデータ
            hol_new.append(int(one_hol[len(one_hol)-mask_bit-1]))#mask_bit番目の数字を読む
        bits.append(hol_new)
    return bits
def compare(data,num_list):#num_listはみたいマスクの階層
    data_list=[]
    for num in num_list:
        data_=read(data,num)
        data_list.append(np.array(data_))
    result_string = " | ".join([f"data_list[{i}]" for i in range(len(data_list))])
    print(result_string)
    return eval(result_string),data_list
```

```python
from astropy.modeling import models
from astropy.table import Table
from astropy.modeling import fitting

def fit_star(data, x0, y0, box_size=15):
    ny, nx = data.shape
    x_min = int(max(x0 - box_size//2, 0))
    x_max = int(min(x0 + box_size//2 + 1, nx))
    y_min = int(max(y0 - box_size//2, 0))
    y_max = int(min(y0 + box_size//2 + 1, ny))

    cutout = data[y_min:y_max, x_min:x_max]
    yy, xx = np.mgrid[y_min:y_max, x_min:x_max]

    # 初期モデル
    init_model = models.Gaussian2D(
        amplitude=cutout.max(),
        x_mean=x0,
        y_mean=y0,
        x_stddev=2,
        y_stddev=2,
        theta=0
    )

    fitter = fitting.LevMarLSQFitter()
    fitted_model = fitter(init_model, xx, yy, cutout)

    return fitted_model

# ----------------------------
# 📋 全星に対してループ
# ----------------------------
primary = det_smooth[det_smooth.deblended_flag==0]
results = []
positions = Table()
positions['x'] = primary['xcentroid']
positions['y'] = primary['ycentroid']
for row in tqdm(positions):
    x0, y0 = row['x'], row['y']
    model = fit_star(data_sub, x0, y0)

    results.append({
        'x_fit': model.x_mean.value,
        'y_fit': model.y_mean.value,
        'amplitude': model.amplitude.value,
        'x_stddev': model.x_stddev.value,
        'y_stddev': model.y_stddev.value,
        'theta_deg': np.degrees(model.theta.value)
    })

# 結果をテーブルに
results_table = Table(rows=results)
primary['x_fit']=results_table['x_fit']
primary['y_fit']=results_table['y_fit']
primary['amplitude']=results_table['amplitude']
primary['x_stddev']=results_table['x_stddev']
primary['y_stddev']=results_table['y_stddev']
primary['theta_deg']=results_table['theta_deg']
```

```python
import numpy as np
from numba import njit
ny, nx = data_sub.shape
y_grid, x_grid = np.mgrid[0:ny, 0:nx]
x_flat = x_grid.ravel()
y_flat = y_grid.ravel()
@njit(nopython=True)
def add_gaussian(model_flat, x_flat, y_flat, x0, y0, sx, sy, theta, amp):
    """
    1つの2Dガウシアンを加算する
    """
    ct = np.cos(theta)
    st = np.sin(theta)
    for i in range(x_flat.size):
         x_shift = x_flat[i] - x0
         y_shift = y_flat[i] - y0
        
         x_rot =  ct * x_shift + st * y_shift
         y_rot = -st * x_shift + ct * y_shift
        
         exponent = -0.5 * ((x_rot / sx)**2 + (y_rot / sy)**2)
         model_flat[i] += amp * np.exp(exponent)

    return model_flat
model_flat = np.zeros_like(x_flat, dtype=float)

for row in tqdm(results):
    x0 = row['x_fit']
    y0 = row['y_fit']
    sx = row['x_stddev']
    sy = row['y_stddev']
    theta = np.radians(row['theta_deg'])
    amp = row['amplitude']

    model_flat = add_gaussian(model_flat, x_flat, y_flat, x0, y0, sx, sy, theta, amp)

model_image = model_flat.reshape(ny, nx)
```

```python
from astropy.modeling import models, fitting
import numpy as np
cat = SourceCatalog(data_sub, segm_deblend)
def moment_to_shape(cov):
    """
    共分散行列から
    major_sigma, minor_sigma, ellipticity, theta を計算
    """
    cov = np.array(cov)
    if hasattr(cov[0, 0], 'value'):  # Quantity の場合
        cov = cov.value

    eigvals, eigvecs = np.linalg.eigh(cov)

    idx = np.argsort(eigvals)[::-1]
    eigvals = eigvals[idx]
    eigvecs = eigvecs[:, idx]

    major_sigma = np.sqrt(eigvals[0])
    minor_sigma = np.sqrt(eigvals[1])

    ellipticity = 1 - (minor_sigma / major_sigma)
    theta = np.degrees(np.arctan2(eigvecs[1, 0], eigvecs[0, 0]))

    return float(major_sigma), float(minor_sigma), float(ellipticity), float(theta)

from astropy.modeling import models, fitting
import numpy as np

def fit_elliptical_gaussian(data, x0, y0, cov, box_size=30):
    """
    画像 data 上の (x0, y0) 位置にある天体に 2D 楕円ガウスをフィット。
    共分散行列 cov から初期パラメータを推定。
    """
    ny, nx = data.shape
    half = box_size // 2

    xmin = max(int(x0) - half, 0)
    xmax = min(int(x0) + half + 1, nx)
    ymin = max(int(y0) - half, 0)
    ymax = min(int(y0) + half + 1, ny)

    cutout = data[ymin:ymax, xmin:xmax]

    if cutout.size == 0 or cutout.shape[0] < 3 or cutout.shape[1] < 3:
        return None  # スキップ

    y, x = np.mgrid[0:cutout.shape[0], 0:cutout.shape[1]]
    y_rel = y - (ymax - ymin) / 2
    x_rel = x - (xmax - xmin) / 2

    # 共分散から形状推定
    eigvals, eigvecs = np.linalg.eigh(cov)
    major_sigma = np.sqrt(np.max(eigvals))
    minor_sigma = np.sqrt(np.min(eigvals))
    theta = np.degrees(np.arctan2(eigvecs[1, 1], eigvecs[0, 1]))  # 方向角
    theta_rad = np.radians(theta)

    # 初期モデル
    g_init = models.Gaussian2D(
        amplitude=cutout.max(),
        x_mean=0, y_mean=0,
        x_stddev=major_sigma,
        y_stddev=minor_sigma,
        theta=theta_rad
    )

    fitter = fitting.LevMarLSQFitter()
    g_fit = fitter(g_init, x_rel, y_rel, cutout)

    flux = 2 * np.pi * g_fit.amplitude * g_fit.x_stddev * g_fit.y_stddev

    result = {
        "amplitude": g_fit.amplitude.value,
        "x_mean": g_fit.x_mean.value + x0,
        "y_mean": g_fit.y_mean.value + y0,
        "x_stddev": g_fit.x_stddev.value,
        "y_stddev": g_fit.y_stddev.value,
        "theta": np.degrees(g_fit.theta.value),
        "flux": flux
    }
    return result

```

```python
import numpy as np
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.stats import sigma_clipped_stats
from astropy.modeling.fitting import LevMarLSQFitter
from astropy.convolution import convolve
from astropy import units as u

from photutils.background import Background2D, MMMBackground, LocalBackground
from photutils.segmentation import detect_sources, SourceCatalog
from photutils.psf import IterativePSFPhotometry, CircularGaussianPRF
from photutils.detection import DAOStarFinder

# -------------------------------
# 画像読み込み & 背景推定
data = calexp[1].data

bkg = Background2D(data, box_size=50, filter_size=3, bkg_estimator=MedianBackground())
data_sub = data - bkg.background

# -------------------------------
# PSF（簡易ガウシアン）でスムージング
sigma = 1
fwhm = 2.3548 * sigma
psf_model = CircularGaussianPRF(fwhm=fwhm)  # 1.5 pixel
y, x = np.mgrid[-5:6, -5:6]
psf = psf_model(x, y)
psf /= np.sum(psf)

smoothed = convolve(data_sub, psf)  # これでOK

# -------------------------------
# ピーク検出
threshold = 3. * bkg.background_rms_median
segm = detect_sources(smoothed, threshold, npixels=5)

fig = plt.figure(figsize=(10, 10))
plt.imshow(segm.data, origin='lower', cmap='viridis')
plt.title("Detected Sources")
plt.colorbar()
plt.show()

# -------------------------------
# 検出した領域の重心座標
cat = SourceCatalog(data_sub, segm)
init_x = cat.xcentroid
init_y = cat.ycentroid
positions = np.transpose([init_x, init_y])

# -------------------------------
# PSF 測光
sigma = 1.5
fwhm = 2.3548 * sigma
fit_shape = (5, 5)
finder = DAOStarFinder(10.0, 2.0)
bkgstat = MMMBackground()
localbkg_estimator = LocalBackground(5, 10, bkgstat)
psf_model = CircularGaussianPRF(fwhm=fwhm)

# PSF 測光オブジェクト
photometry = IterativePSFPhotometry(psf_model, fit_shape, finder=finder,
                                  localbkg_estimator=localbkg_estimator,
                                  aperture_radius=4)


result_tab = photometry(data_sub)

print(photometry)

# 結果表示
plt.figure(figsize=(10, 10))
plt.imshow(data_sub, cmap='gray', origin='lower', vmin=np.percentile(data_sub, 5), vmax=np.percentile(data_sub, 99))
plt.plot(result_tab['x_fit'], result_tab['y_fit'], 'rx', markersize=10, label='PSF Photometry')
plt.legend()
plt.show()
```
